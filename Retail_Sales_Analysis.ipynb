```python
import pandas as pd

df = pd.read_csv(
    '/content/retail_data/Sample - Superstore.csv',
    encoding='latin1'
)

print("Dataset loaded!")
print("Shape:", df.shape)
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

df['Shipping Days'] = (df['Ship Date'] - df['Order Date']).dt.days
df['Order Year'] = df['Order Date'].dt.year
df['Order Month'] = df['Order Date'].dt.month
df['Month Name'] = df['Order Date'].dt.month_name()
df['Profit Margin'] = (df['Profit'] / df['Sales']) * 100

print("Ready!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
yearly_analysis = df.groupby('Order Year').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum')
)

yearly_analysis
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    yearly_analysis.index,
    yearly_analysis['Total_Sales'],
    marker='o'
)

plt.title('Yearly Sales Trend')
plt.xlabel('Year')
plt.ylabel('Total Sales')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import sqlite3
import pandas as pd

# Create database
connection = sqlite3.connect('/content/retail_sales.db')

# Create sales table from our DataFrame
df.to_sql(
    'sales',
    connection,
    if_exists='replace',
    index=False
)

print("✅ SQL database created!")
print("✅ Table 'sales' created!")
```

```python
query = """
SELECT *
FROM sales
LIMIT 5;
"""

result = pd.read_sql_query(query, connection)

result
```

```python
query = """
SELECT
    ROUND(SUM(Sales), 2) AS Total_Sales
FROM sales;
"""

pd.read_sql_query(query, connection)
```

```python
query = """
SELECT
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM sales;
"""

pd.read_sql_query(query, connection)
```

```python
query = """
SELECT
    Category,
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM sales
GROUP BY Category
ORDER BY Total_Sales DESC;
"""

pd.read_sql_query(query, connection)
```

```python
query = """
SELECT
    "Product Name",
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM sales
GROUP BY "Product Name"
ORDER BY Total_Sales DESC
LIMIT 10;
"""

pd.read_sql_query(query, connection)
```

```python
query = """
SELECT
    "Product Name",
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM sales
GROUP BY "Product Name"
ORDER BY Total_Profit DESC
LIMIT 10;
"""

pd.read_sql_query(query, connection)
```

```python
query = """
SELECT
    "Product Name",
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM sales
GROUP BY "Product Name"
HAVING SUM(Profit) < 0
ORDER BY Total_Profit ASC
LIMIT 10;
"""

pd.read_sql_query(query, connection)
```

```python
query = """
SELECT
    "Customer ID",
    "Customer Name",
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM sales
GROUP BY "Customer ID", "Customer Name"
ORDER BY Total_Profit DESC
LIMIT 10;
"""

pd.read_sql_query(query, connection)
```

```python
query = """
SELECT
    State,
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM sales
GROUP BY State
ORDER BY Total_Profit DESC
LIMIT 10;
"""

pd.read_sql_query(query, connection)
```

```python
query = """
SELECT
    State,
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM sales
GROUP BY State
ORDER BY Total_Profit ASC
LIMIT 10;
"""

pd.read_sql_query(query, connection)
```

```python
query = """
SELECT
    Discount,
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit,
    ROUND(
        (SUM(Profit) * 100.0) / SUM(Sales),
        2
    ) AS Profit_Margin_Percent
FROM sales
GROUP BY Discount
ORDER BY Discount;
"""

pd.read_sql_query(query, connection)
```

```python
query = """
SELECT
    "Order Year" AS Year,
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM sales
GROUP BY "Order Year"
ORDER BY Year;
"""

pd.read_sql_query(query, connection)
```

```python
query = """
SELECT
    Category,
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit,
    RANK() OVER (
        ORDER BY SUM(Profit) DESC
    ) AS Profit_Rank
FROM sales
GROUP BY Category;
"""

pd.read_sql_query(query, connection)
```

```python
monthly_analysis = df.groupby('Order Month').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum')
)

print(monthly_analysis)
```

```python
customer_sales = df.groupby(
    ['Customer ID', 'Customer Name']
).agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Total_Orders=('Order ID', 'nunique')
).sort_values('Total_Sales', ascending=False)

print(customer_sales.head(10))
```

```python
subcategory_analysis = df.groupby('Sub-Category').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Total_Quantity=('Quantity', 'sum')
).sort_values('Total_Profit', ascending=False)

print(subcategory_analysis)
```

```python
discount_analysis = df.groupby('Discount').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Total_Quantity=('Quantity', 'sum'),
    Number_of_Orders=('Order ID', 'nunique')
).sort_index()

discount_analysis['Profit_Margin_%'] = (
    discount_analysis['Total_Profit'] /
    discount_analysis['Total_Sales']
) * 100

print(discount_analysis)
```

```python
product_analysis = df.groupby('Product Name').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Total_Quantity=('Quantity', 'sum')
)

product_analysis['Profit_Margin_%'] = (
    product_analysis['Total_Profit'] /
    product_analysis['Total_Sales']
) * 100

high_sales_loss = product_analysis[
    (product_analysis['Total_Sales'] >
     product_analysis['Total_Sales'].quantile(0.75)) &
    (product_analysis['Total_Profit'] < 0)
].sort_values('Total_Sales', ascending=False)

print(high_sales_loss)
```

```python
state_analysis = df.groupby('State').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Total_Quantity=('Quantity', 'sum')
).sort_values('Total_Sales', ascending=False)

print(state_analysis.head(10))
```

```python
city_analysis = df.groupby('City').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Total_Orders=('Order ID', 'nunique')
).sort_values('Total_Sales', ascending=False)

print(city_analysis.head(10))
```

```python
ship_mode_analysis = df.groupby('Ship Mode').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Average_Shipping_Days=('Shipping Days', 'mean'),
    Total_Orders=('Order ID', 'nunique')
).sort_values('Total_Sales', ascending=False)

ship_mode_analysis['Average_Shipping_Days'] = (
    ship_mode_analysis['Average_Shipping_Days'].round(2)
)

print(ship_mode_analysis)
```

```python
segment_analysis = df.groupby('Segment').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Total_Quantity=('Quantity', 'sum'),
    Total_Orders=('Order ID', 'nunique')
).sort_values('Total_Sales', ascending=False)

print(segment_analysis)
```

```python
category_region = pd.pivot_table(
    df,
    values='Profit',
    index='Region',
    columns='Category',
    aggfunc='sum'
)

print(category_region)
```

```python
kpi_summary = pd.DataFrame({
    'KPI': [
        'Total Sales',
        'Total Profit',
        'Total Orders',
        'Total Quantity',
        'Average Order Value',
        'Overall Profit Margin'
    ],
    'Value': [
        df['Sales'].sum(),
        df['Profit'].sum(),
        df['Order ID'].nunique(),
        df['Quantity'].sum(),
        df['Sales'].sum() / df['Order ID'].nunique(),
        (df['Profit'].sum() / df['Sales'].sum()) * 100
    ]
})

kpi_summary['Value'] = kpi_summary['Value'].round(2)

print(kpi_summary)
```

```python
query = """
SELECT
    "Product Name",
    SUM(Sales) AS Total_Sales
FROM sales
GROUP BY "Product Name"
ORDER BY Total_Sales DESC
LIMIT 10;
"""

pd.read_sql_query(query, connection)
```

```python
query = """
SELECT
    "Product Name",
    SUM(Profit) AS Total_Profit
FROM sales
GROUP BY "Product Name"
ORDER BY Total_Profit DESC
LIMIT 10;
"""

pd.read_sql_query(query, connection)
```

```python
query = """
SELECT
    "Product Name",
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM sales
GROUP BY "Product Name"
HAVING Total_Profit < 0
ORDER BY Total_Profit ASC
LIMIT 10;
"""

pd.read_sql_query(query, connection)
```

```python
query = """
SELECT
    "Customer Name",
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM sales
GROUP BY "Customer Name"
ORDER BY Total_Profit DESC
LIMIT 10;
"""

pd.read_sql_query(query, connection)
```

```python
query = """
SELECT
    State,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM sales
GROUP BY State
ORDER BY Total_Profit DESC;
"""

pd.read_sql_query(query, connection)
```

```python
query = """
SELECT
    Discount,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM sales
GROUP BY Discount
ORDER BY Discount;
"""

pd.read_sql_query(query, connection)
```

```python
query = """
SELECT
    "Order Year",
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM sales
GROUP BY "Order Year"
ORDER BY "Order Year";
"""

pd.read_sql_query(query, connection)
```

```python
query = """
SELECT
    Segment,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit,
    SUM(Quantity) AS Total_Quantity
FROM sales
GROUP BY Segment
ORDER BY Total_Sales DESC;
"""

pd.read_sql_query(query, connection)
```

```python
query = """
SELECT
    "Ship Mode",
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit,
    AVG("Shipping Days") AS Average_Shipping_Days
FROM sales
GROUP BY "Ship Mode"
ORDER BY Total_Sales DESC;
"""

pd.read_sql_query(query, connection)
```
